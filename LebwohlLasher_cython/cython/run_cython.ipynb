```python
from LebwohlLasher_cython import main

#params
ITERATIONS = [500, 1000, 2000,5000]
SIZES = [20, 50, 100]
TEMP = 0.5
PLOTFLAG = 0

PYTHON_EXEC = "python"
MODULE_NAME = "LebwohlLasher_cython"   


#run all param combinations
for size in SIZES:
    for iters in ITERATIONS:

        print(f"Running Size={size}, Steps={iters}")

        cmd = [
            PYTHON_EXEC,
            "-m",
            MODULE_NAME,
            str(iters),
            str(size),
            str(TEMP),
            str(PLOTFLAG),
        ]

        main("LebwohlLasher", iters, size, TEMP, PLOTFLAG)
```
