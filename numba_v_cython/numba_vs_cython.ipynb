```python
from itertools import product
from LebwohlLasher_cython.cython.LebwohlLasher_cython import main


def run_model(module, steps, size, temp=0.5, plotflag=0):
    cmd = ["python", "-m", module, str(steps), str(size), str(temp), str(plotflag)]
    print(" ".join(cmd))
    subprocess.run(cmd, check=False)

MODELS = {
    # "base": "LebwohlLasher",
    # "numba": "LebwohlLasher_numba.LebwohlLasher_numba",
    "cython": "LebwohlLasher_cython.cython.LebwohlLasher_cython",
}

SIZES = [20, 50, 100]
STEPS = [100, 500, 1000]
TEMP = 0.5


for name, mod in MODELS.items():
    for nmax, nsteps in product(SIZES, STEPS):
        print(f"Run | {name} | N={nmax} steps={nsteps}")
        run_model(mod, nsteps, nmax, TEMP, plotflag=0)


for size in SIZES:
    for iters in STEPS:
        print(f"Running Cython | N={size} | Steps={iters}")
        main("LebwohlLasher_cython", iters, size, TEMP, 0)
```

```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt

# 1) Move into results folder
#RESULTS_DIR = "numba_v_cython"
#os.chdir(RESULTS_DIR)
print("Now in:", os.getcwd())

# 2) Define file patterns for each implementation (edit if needed)
PREFIXES = {
    "base": "LL-*",
    "numba": "Nb-O*",
    "cython": "cython-*",   # change to your actual cython prefix if different
}

# 3) Parse one output file (assumes same header structure as your baseline-style output)
def parse_file(fname):
    with open(fname) as f:
        lines = f.readlines()

    # These indices match your earlier parsing logic
    size_line = lines[2]
    steps_line = lines[3]
    runtime_line = lines[5]

    nmax = int(size_line.split(":")[1].split("x")[0].strip())
    nsteps = int(steps_line.split(":")[1].strip())
    runtime = float(runtime_line.split(":")[1].strip())
    return nmax, nsteps, runtime

# 4) Build df_all
rows = []
for impl, pattern in PREFIXES.items():
    for fname in glob.glob(pattern):
        try:
            nmax, nsteps, runtime = parse_file(fname)
            rows.append([impl, fname, nmax, nsteps, runtime])
        except Exception as e:
            print(f"Skipping {fname}: {e}")

df_all = pd.DataFrame(rows, columns=["implementation", "file", "nmax", "nsteps", "runtime"])
df_all = df_all.sort_values(["implementation", "nmax", "nsteps"])
display(df_all.head())

# 5) Summary stats + speedup vs baseline
summary = (
    df_all.groupby(["implementation", "nmax", "nsteps"])["runtime"]
    .agg(runtime_mean="mean", runtime_sd="std", n="count")
    .reset_index()
)

base_lookup = summary[summary["implementation"] == "base"][["nmax", "nsteps", "runtime_mean"]].rename(
    columns={"runtime_mean": "base_runtime_mean"}
)

summary = summary.merge(base_lookup, on=["nmax", "nsteps"], how="left")
summary["speedup_vs_base"] = summary["base_runtime_mean"] / summary["runtime_mean"]

summary = summary.sort_values(["nmax", "nsteps", "implementation"])
display(summary.head(20))

# 6) Plot A: Runtime vs Steps (pick one N)
N_FIXED = 100
sub = summary[summary["nmax"] == N_FIXED].copy()

fig, ax = plt.subplots()
for impl, g in sub.groupby("implementation"):
    g = g.sort_values("nsteps")
    ax.plot(g["nsteps"], g["runtime_mean"], marker="o", label=impl)

ax.set_xlabel("Steps")
ax.set_ylabel("Runtime (s)")
ax.set_title(f"Runtime vs Steps (N={N_FIXED})")
ax.legend()
plt.tight_layout()
plt.show()

# 7) Plot B: Runtime vs Size (pick one steps)
STEPS_FIXED = 1000
sub = summary[summary["nsteps"] == STEPS_FIXED].copy()

fig, ax = plt.subplots()
for impl, g in sub.groupby("implementation"):
    g = g.sort_values("nmax")
    ax.plot(g["nmax"], g["runtime_mean"], marker="o", label=impl)

ax.set_xlabel("Lattice size N")
ax.set_ylabel("Runtime (s)")
ax.set_title(f"Runtime vs Size (steps={STEPS_FIXED})")
ax.legend()
plt.tight_layout()
plt.show()
```
